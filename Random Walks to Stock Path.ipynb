Task 1.1 - simulate a random walk with 100 steps

```python
import random
import matplotlib.pyplot as plt

def walk():
    pos = 0
    steps = [0] # start at 0
    for i in range(99):
        x = random.choice([1,-1])
        pos += x
        steps.append(pos)
    plt.plot(steps)
plt.show(walk())
```

task 1.2 - simulate 100 random walks with 100 steps each on the same set of axis

```python
for j in range(100):
    walk()
plt.show()
```

Task 2 - simulate 10,000 random walks each wit 400 steps, record the final positions, plot them in a histogram using plt.hist([], bins = ), calculate mean and standard deviation of the final positions.

Prediction: each step has variance 1, and the variance of independant steps should add up. therefore, for n steps, the std should be sqrt(n), furthermore, the plot should roughly be a normal distribution.

result: the std was 19.8 which is approx. 20, i.e. sqrt(400) which proves the prediciton to be true. moreover, the plot for the final positions is similar to a normal distribution.

```python
import numpy

positions = []
for j in range(10000):
    position = 0
    for i in range(400):
        x = random.choice([1,-1])
        position += x
    positions.append(position)

plt.hist(positions, bins = 30)
plt.hist(positions, bins = range(-41,42,2))
plt.show()
print(numpy.mean(positions))
print(numpy.std(positions))
# pred - std is 20
```

Task 3 - a simulated stock price.
1- Simulate one trading year (252 days) of the stock price, keeping the price on every day.
2- Plot 50 independent simulated years on the same axes.
3- Change the drift and the volatility, one at a time, and describe what each does to the picture.

```python
import random
import numpy as np
import matplotlib.pyplot as plt

finals = []
for day in range(50):
    price = 100
    day = [100]
    for i in range(252):
        z = random.gauss(0,1)
        change = 1 + 0.0005 + 0.01*z
        price *= change
        day.append(price)
    finals.append(day[-1])
    plt.plot(day)
plt.show()

plt.hist(finals, bins = 30)
print("mean: " + str(np.mean(finals)))
print("std: " + str(np.std(finals)))
```

Task 4 - build on this, do 10,000 paths and find the proportion of prices which enc up more than 110.

```python
import random
import numpy as np
import matplotlib.pyplot as plt

finals = []
count = 0
for day in range(10000):
    price = 100
    for i in range(252):
        z = random.gauss(0,1)
        change = 1 + 0.0005 + 0.01*z
        price *= change
    finals.append(price)
for value in finals:
    if value >= 110:
        count += 1
print(count/len(finals))
```
